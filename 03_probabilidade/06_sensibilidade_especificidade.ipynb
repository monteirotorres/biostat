# Sensibilidade e especificidade

São propriedades **do teste**, calculadas em pessoas cujo diagnóstico verdadeiro é conhecido.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
df = pd.DataFrame({
    'doente':         [True]*90  + [False]*910,
    'teste_positivo': [True]*85  + [False]*5   + [True]*45 + [False]*865,
})
tab = pd.crosstab(df['doente'], df['teste_positivo'])
tab
```

```python
VP, FN = tab.loc[True, True], tab.loc[True, False]
FP, VN = tab.loc[False, True], tab.loc[False, False]
print('Sensibilidade:', VP / (VP + FN))
print('Especificidade:', VN / (VN + FP))
```
