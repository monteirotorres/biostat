# Valor preditivo positivo e negativo

Respondem à pergunta do paciente: 'meu teste deu positivo, qual a chance de eu ter mesmo a doença?'

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
df = pd.DataFrame({
    'doente':         [True]*90  + [False]*910,
    'teste_positivo': [True]*85  + [False]*5   + [True]*45 + [False]*865,
})
tab = pd.crosstab(df['teste_positivo'], df['doente'])
tab
```

```python
VP = tab.loc[True, True]
FP = tab.loc[True, False]
VN = tab.loc[False, False]
FN = tab.loc[False, True]

print('VPP:', VP / (VP + FP))
print('VPN:', VN / (VN + FN))
```

VPP e VPN **dependem da prevalência**, ao contrário de sensibilidade e especificidade.
