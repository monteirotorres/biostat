# Matriz de confusão

Tabela 2×2 que mostra acertos e erros do teste.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
import numpy as np
rng = np.random.default_rng(0)
doente = rng.random(1000) < 0.20
teste_pos = np.where(doente, rng.random(1000) < 0.90, rng.random(1000) < 0.10)
```

```python
matriz = pd.crosstab(pd.Series(doente, name='verdadeiro'),
                     pd.Series(teste_pos, name='teste'))
matriz
```

```python
sns.heatmap(matriz, annot=True, fmt='d', cmap='Blues');
```
