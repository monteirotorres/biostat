# Distribuição binomial

Conta o número de sucessos em $n$ tentativas independentes, cada uma com a mesma probabilidade $p$ de sucesso.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## Lançar 10 moedas — quantas caras?

```python
import numpy as np
n, p = 10, 0.5
k = np.arange(0, n+1)
probs = stats.binom.pmf(k, n, p)
pd.Series(probs, index=k).plot.bar()
plt.xlabel('caras')
plt.ylabel('probabilidade');
```

## Pelo menos 7 caras em 10 jogadas

```python
1 - stats.binom.cdf(6, n=10, p=0.5)
```

## Simulando 5000 experimentos

```python
amostras = pd.Series(stats.binom.rvs(n=10, p=0.5, size=5000, random_state=0))
amostras.value_counts().sort_index()
```
