# Testes binários

Um teste binário retorna apenas dois resultados: **positivo** ou **negativo**. Quase todo teste diagnóstico cai aqui.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

Vamos simular 1000 pacientes em que 10% têm a doença e o teste acerta 95% das vezes.

```python
import numpy as np
rng = np.random.default_rng(0)
n = 1000
doente = rng.random(n) < 0.10
# Sensibilidade = 95%, Especificidade = 95%
teste_pos = np.where(doente,
                     rng.random(n) < 0.95,
                     rng.random(n) < 0.05)
df = pd.DataFrame({'doente': doente, 'teste_positivo': teste_pos})
df.head()
```

```python
pd.crosstab(df['doente'], df['teste_positivo'], margins=True)
```
