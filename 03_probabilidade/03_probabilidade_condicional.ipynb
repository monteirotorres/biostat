# Probabilidade condicional

$P(A \mid B)$ = probabilidade de A, **sabendo que** B aconteceu.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
df = sns.load_dataset('titanic')
df[['sex', 'survived']].head()
```

## P(sobreviver)

```python
df['survived'].mean()
```

## P(sobreviver | sexo = feminino)

```python
df.groupby('sex')['survived'].mean()
```

Repare como a probabilidade muda muito quando condicionamos no sexo. Os dois eventos não são independentes.
