# Curva ROC

Mostra como o teste se comporta para **diferentes valores de corte**.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
import numpy as np
rng = np.random.default_rng(0)
n = 500
doentes = stats.norm.rvs(loc=60, scale=10, size=n, random_state=0)
saudaveis = stats.norm.rvs(loc=40, scale=10, size=n, random_state=1)
df = pd.concat([
    pd.DataFrame({'valor': doentes,   'doente': True}),
    pd.DataFrame({'valor': saudaveis, 'doente': False}),
])
```

```python
sns.histplot(data=df, x='valor', hue='doente', bins=30);
```

```python
# Calcula sens. e (1-esp.) para cada corte
cortes = pd.Series(range(20, 80))
tpr = cortes.apply(lambda c: ((df['valor'] >= c) & df['doente']).sum() / df['doente'].sum())
fpr = cortes.apply(lambda c: ((df['valor'] >= c) & ~df['doente']).sum() / (~df['doente']).sum())
plt.plot(fpr, tpr)
plt.plot([0,1], [0,1], '--', color='gray')
plt.xlabel('1 - especificidade (FPR)')
plt.ylabel('sensibilidade (TPR)');
```
