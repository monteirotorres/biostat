Quantas formas existem de organizar ou escolher elementos?

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## Fatorial

```python
from math import factorial
factorial(5)  # 5! = 120
```

## Combinações: escolher k entre n (a ordem não importa)

```python
from math import comb
comb(60, 6)  # mega-sena
```

## Permutações: ordenar k entre n (a ordem importa)

```python
from math import perm
perm(10, 3)  # quantos pódios em uma corrida de 10 atletas?
```

## Probabilidade de ganhar na mega-sena

```python
1 / comb(60, 6)
```
