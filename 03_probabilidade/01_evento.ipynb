# Evento — definição

Um **evento** é um resultado (ou conjunto de resultados) possível de um experimento aleatório.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## Simulando 10.000 lançamentos de moeda

```python
import numpy as np
lancamentos = pd.Series(np.random.default_rng(0).choice(['cara', 'coroa'], size=10_000))
lancamentos.value_counts(normalize=True)
```

## Simulando lançamentos de dado

```python
dado = pd.Series(np.random.default_rng(0).integers(1, 7, size=10_000))
(dado == 6).mean()  # P(sair 6)
```

```python
(dado % 2 == 0).mean()  # P(número par)
```
