# Notação científica

Python escreve automaticamente em notação científica para números muito grandes ou pequenos.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
# Número de Avogadro
n_avogadro = 6.022e23
print(n_avogadro)
```

```python
# Massa de um próton (kg)
m_proton = 1.6726e-27
print(m_proton)
```

## Convertendo automaticamente

```python
df = pd.DataFrame({'concentracao_M': [1e-3, 5e-6, 2.3e-9, 4.5e-12]})
df
```

```python
# Forçando notação científica na exibição
pd.set_option('display.float_format', '{:.2e}'.format)
df
```
