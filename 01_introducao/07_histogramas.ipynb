# Histogramas

O histograma mostra a **forma** da distribuição dos dados.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
df = sns.load_dataset('penguins').dropna()
```

## Histograma simples

```python
sns.histplot(data=df, x='flipper_length_mm', bins=20);
```

## Mudando o número de barras

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 3))
for ax, b in zip(axes, [5, 15, 50]):
    sns.histplot(data=df, x='flipper_length_mm', bins=b, ax=ax)
    ax.set_title(f'{b} barras')
```

Poucas barras = perdemos detalhes. Muitas barras = ruído. O ideal fica no meio do caminho.

## Separando por grupo

```python
sns.histplot(data=df, x='flipper_length_mm', hue='species', bins=20);
```

## Densidade (forma suavizada)

```python
sns.kdeplot(data=df, x='flipper_length_mm', hue='species', fill=True);
```
