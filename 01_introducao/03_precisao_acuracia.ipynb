# Precisão e acurácia

Vamos simular quatro situações para visualizar a diferença.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
# Valor verdadeiro
verdadeiro = 100

cenarios = {
    'Acurado e preciso':       stats.norm.rvs(loc=100, scale=1,  size=50, random_state=0),
    'Preciso, não acurado':    stats.norm.rvs(loc=110, scale=1,  size=50, random_state=1),
    'Acurado, não preciso':    stats.norm.rvs(loc=100, scale=8,  size=50, random_state=2),
    'Nem acurado, nem preciso':stats.norm.rvs(loc=110, scale=8,  size=50, random_state=3),
}
df = pd.DataFrame(cenarios)
df.head()
```

```python
sns.boxplot(data=df)
plt.axhline(verdadeiro, color='red', linestyle='--', label='valor verdadeiro')
plt.xticks(rotation=30, ha='right')
plt.legend();
```

- **Acurácia** = quão perto da linha vermelha (valor verdadeiro).
- **Precisão** = quão pequena é a caixa (pouca dispersão).
