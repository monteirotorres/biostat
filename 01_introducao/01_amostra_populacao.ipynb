# Amostra e população

Vamos simular uma população e tirar amostras dela para ver como cada amostra estima o parâmetro verdadeiro.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## Criando uma 'população' simulada

Vamos imaginar a altura (em cm) de **100.000 adultos**. Como é uma simulação, nós conhecemos os parâmetros verdadeiros.

```python
populacao = pd.Series(stats.norm.rvs(loc=170, scale=8, size=100_000, random_state=0))
populacao.describe()
```

A média **verdadeira** da população é 170 cm e o desvio padrão é 8 cm.

## Tirando uma amostra

Vamos sortear apenas 30 pessoas dessa população.

```python
amostra = populacao.sample(n=30, random_state=1)
amostra.describe()
```

Repare que a média da amostra (`mean`) é **próxima**, mas não igual a 170. Isso é normal — chamamos esse fenômeno de **erro amostral**.

## E se tirarmos várias amostras?

```python
medias = [populacao.sample(n=30, random_state=i).mean() for i in range(500)]
medias = pd.Series(medias)
sns.histplot(medias, bins=30)
plt.axvline(170, color='red', label='média da população')
plt.xlabel('média da amostra')
plt.legend();
```

Cada amostra dá uma média um pouco diferente, mas todas se concentram em torno do valor verdadeiro. Essa é a ideia central da inferência estatística.
