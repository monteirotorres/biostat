# Descrição vs. inferência

A **descrição** resume os dados que temos em mãos. A **inferência** generaliza esses dados para uma população maior.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## Carregando um conjunto de dados real

Vamos usar o famoso dataset de pinguins do `seaborn`.

```python
df = sns.load_dataset('penguins').dropna()
df.head()
```

## Descrição: o que esses 333 pinguins têm?

```python
df['body_mass_g'].describe()
```

Esse é um resumo **descritivo** — fala apenas dos pinguins que estão no dataset.

## Inferência: o que isso diz sobre TODOS os pinguins?

```python
stats.t.interval(0.95,
                df=len(df) - 1,
                loc=df['body_mass_g'].mean(),
                scale=stats.sem(df['body_mass_g']))
```

Esse intervalo é uma **inferência**: estimamos que a massa média dos pinguins (em geral) está dentro desse intervalo, com 95% de confiança.
