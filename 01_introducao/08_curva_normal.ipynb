# Curva normal

A distribuição normal é totalmente definida por dois parâmetros: **média** ($\mu$) e **desvio padrão** ($\sigma$).

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## Como muda com $\mu$ e $\sigma$?

```python
import numpy as np
x = np.linspace(-15, 15, 500)
for mu, sigma, label in [(0, 1, 'μ=0, σ=1'),
                         (0, 3, 'μ=0, σ=3'),
                         (5, 1, 'μ=5, σ=1')]:
    plt.plot(x, stats.norm.pdf(x, mu, sigma), label=label)
plt.legend();
```

## A regra 68-95-99,7

Vamos verificar empiricamente em uma amostra de 100.000 valores.

```python
amostra = pd.Series(stats.norm.rvs(loc=0, scale=1, size=100_000, random_state=0))
for k in [1, 2, 3]:
    pct = ((amostra > -k) & (amostra < k)).mean() * 100
    print(f'Entre -{k}σ e +{k}σ: {pct:.1f}%')
```

Bem próximo de 68%, 95% e 99,7%, como prevê a teoria.

## Calculando probabilidades

```python
# P(X < 1.5) em uma normal padrão
stats.norm.cdf(1.5, loc=0, scale=1)
```

```python
# P(-1 < X < 1)
stats.norm.cdf(1) - stats.norm.cdf(-1)
```
