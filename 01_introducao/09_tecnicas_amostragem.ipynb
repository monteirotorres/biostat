# Técnicas de amostragem

Como o método de amostragem afeta a representatividade.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## Uma população com dois estratos

```python
rng = np.random.default_rng(0)
pop = pd.DataFrame({
    'grupo': ['A']*8000 + ['B']*2000,
    'valor': np.concatenate([rng.normal(50, 5, 8000), rng.normal(80, 5, 2000)])
})
pop.groupby('grupo')['valor'].mean()
```

## Amostra aleatória simples

```python
simples = pop.sample(100, random_state=1)
simples['grupo'].value_counts()
```

## Amostra estratificada (proporcional)

```python
estrat = pop.groupby('grupo', group_keys=False).apply(lambda g: g.sample(frac=0.01, random_state=1))
estrat['grupo'].value_counts()
```

```python
print('média pop:       ', pop['valor'].mean())
print('média simples:   ', simples['valor'].mean())
print('média estratif.: ', estrat['valor'].mean())
```

A estratificada garante a proporção certa de cada grupo e costuma estimar melhor a média.
