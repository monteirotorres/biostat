# Desvio interquartil (IQR)

Diferença entre o terceiro e o primeiro quartil. Descreve a dispersão dos **50% centrais** dos dados — robusto a outliers.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
df = sns.load_dataset('penguins').dropna()
q1 = df['body_mass_g'].quantile(0.25)
q3 = df['body_mass_g'].quantile(0.75)
iqr = q3 - q1
print(f'Q1 = {q1}, Q3 = {q3}, IQR = {iqr}')
```

Função pronta:

```python
stats.iqr(df['body_mass_g'])
```

## Boxplot mostra o IQR visualmente

```python
sns.boxplot(data=df, y='body_mass_g');
```
