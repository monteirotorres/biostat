# Média x mediana

Quando usar qual?

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## Distribuição simétrica → tanto faz

```python
simetrica = pd.Series(stats.norm.rvs(loc=50, scale=10, size=1000, random_state=0))
print('média:  ', simetrica.mean())
print('mediana:', simetrica.median())
sns.histplot(simetrica);
```

## Distribuição assimétrica → mediana é mais representativa

```python
assimetrica = pd.Series(stats.expon.rvs(scale=50, size=1000, random_state=0))
print('média:  ', assimetrica.mean())
print('mediana:', assimetrica.median())
sns.histplot(assimetrica);
```

Em distribuições assimétricas (como renda, tempo de sobrevida, contagens), a **mediana** descreve melhor o 'indivíduo típico'.
