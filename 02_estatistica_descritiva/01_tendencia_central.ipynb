# Medidas de tendência central

As três principais são **moda**, **média** e **mediana**. Cada uma resume os dados de um jeito diferente.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
idades = pd.Series([21, 22, 22, 23, 23, 23, 24, 25, 26, 80])
```

```python
idades.mean(), idades.median(), idades.mode()[0]
```

Repare como o valor `80` puxa a **média** para cima, mas a **mediana** mal se mexe. Essa é a principal lição: cada medida é sensível a coisas diferentes.
