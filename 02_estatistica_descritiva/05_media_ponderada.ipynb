## Exemplo clássico: nota final de uma disciplina

```python
notas = pd.DataFrame({
    'avaliacao': ['P1', 'P2', 'Trabalho', 'Final'],
    'nota':  [7.0, 6.0, 9.0, 8.0],
    'peso':  [2,   3,   1,   4],
})
notas
```

```python
(notas['nota'] * notas['peso']).sum() / notas['peso'].sum()
```
