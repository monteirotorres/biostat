# Variância

Média dos quadrados das distâncias até a média. Por elevar ao quadrado, fica numa unidade diferente (kg² em vez de kg).

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
dados = pd.Series([2, 4, 4, 4, 5, 5, 7, 9])
dados.var()  # divide por n-1 (variância amostral)
```

```python
dados.var(ddof=0)  # divide por n (variância populacional)
```

`ddof=1` (padrão do pandas) é o caso amostral. `ddof=0` é o populacional.
