# Média geométrica

Apropriada para taxas de crescimento, diluições em série, títulos de anticorpos — em geral, dados **multiplicativos**.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
# Crescimento bacteriano (vezes por hora)
taxas = pd.Series([1.5, 2.0, 1.8, 2.2, 1.9])
stats.gmean(taxas)
```

## Diluições em série

Títulos de anticorpos costumam ser reportados como média geométrica.

```python
titulos = pd.Series([1/40, 1/80, 1/160, 1/320, 1/640])
print('Média aritmética:', titulos.mean())
print('Média geométrica:', stats.gmean(titulos))
```
