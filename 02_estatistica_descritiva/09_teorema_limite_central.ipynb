# Teorema do limite central (TLC)

Mesmo que a população tenha uma distribuição estranha, as **médias das amostras** tendem a uma distribuição normal — desde que a amostra seja grande o bastante.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## População nada normal: uma exponencial

```python
import numpy as np
populacao = pd.Series(stats.expon.rvs(scale=10, size=100_000, random_state=0))
sns.histplot(populacao, bins=50);
```

## Tirando 1000 amostras de tamanho n e calculando a média de cada uma

```python
def medias_de_amostras(n, k=1000):
    return pd.Series([populacao.sample(n, random_state=i).mean() for i in range(k)])

fig, axes = plt.subplots(1, 4, figsize=(15, 3))
for ax, n in zip(axes, [2, 5, 30, 100]):
    sns.histplot(medias_de_amostras(n), ax=ax, bins=30)
    ax.set_title(f'n = {n}')
```

Repare: quando a amostra cresce, a distribuição das médias fica cada vez mais parecida com uma curva normal — independentemente da forma original da população. Esse é o TLC.
