# Moda

A **moda** é o valor que aparece mais vezes.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
dados = pd.Series([1, 2, 2, 3, 4, 4, 4, 5])
dados.mode()
```

## Mais de uma moda

```python
dados = pd.Series([1, 2, 2, 3, 3])
dados.mode()  # bimodal
```

## Moda em variáveis categóricas

É a única medida de tendência central que funciona em texto.

```python
df = sns.load_dataset('tips')
df['day'].mode()
```

```python
sns.countplot(data=df, x='day');
```
