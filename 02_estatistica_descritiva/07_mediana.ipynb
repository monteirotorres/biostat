# Mediana

Valor do meio quando os dados estão ordenados.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
dados = pd.Series([3, 1, 4, 1, 5, 9, 2, 6])
dados.median()
```

## Robustez a outliers

```python
d1 = pd.Series([1, 2, 3, 4, 5])
d2 = pd.Series([1, 2, 3, 4, 500])
print('média:    ', d1.mean(), d2.mean())
print('mediana:  ', d1.median(), d2.median())
```

A média muda muito; a mediana mal se mexe.
