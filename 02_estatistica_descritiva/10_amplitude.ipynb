# Amplitude

Maior valor menos o menor valor.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
dados = pd.Series([12, 15, 18, 22, 25, 30])
dados.max() - dados.min()
```

## Limitação

Depende apenas de dois valores e é muito sensível a outliers.

```python
d1 = pd.Series([10, 12, 14, 16, 18])
d2 = pd.Series([10, 12, 14, 16, 500])
print(d1.max() - d1.min())
print(d2.max() - d2.min())
```
