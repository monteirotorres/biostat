# Coeficiente de variação (CV)

Desvio padrão dividido pela média, em porcentagem. É a dispersão **relativa**.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
df = sns.load_dataset('penguins').dropna()
stats.variation(df['body_mass_g']) * 100
```

## Comparando variabilidade entre grupos com escalas diferentes

```python
df.groupby('species')['body_mass_g'].agg(
    media='mean', desvio='std',
    cv=lambda x: stats.variation(x) * 100
)
```
