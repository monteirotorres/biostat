# Desvio padrão

Raiz quadrada da variância. Volta para a mesma unidade dos dados.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
df = sns.load_dataset('penguins').dropna()
df.groupby('species')['body_mass_g'].agg(['mean', 'std'])
```

```python
sns.boxplot(data=df, x='species', y='body_mass_g');
```

## Populacional vs. amostral

```python
dados = pd.Series([10, 12, 14, 16, 18])
print('amostral:    ', dados.std())             # ddof=1 (padrão)
print('populacional:', dados.std(ddof=0))
```
