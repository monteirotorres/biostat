# Exercícios — Tabelas de contingência

```python
import numpy as np
import pandas as pd
from scipy import stats
```

## Exercício 1 — Mortalidade Halotano x Morfina

```python
tab = np.array([[8, 10], [53, 51]])  # mortos / vivos
chi2, p, dof, esp = stats.chi2_contingency(tab)
print('chi2 =', round(chi2,3), '| p =', round(p,4))
print('Fisher:', stats.fisher_exact(tab))
```
