## Exercício 1 — Caixa de produtos

```python
mu = 500 + 50*10
var = 25**2 + 50*2**2
sd = np.sqrt(var)
print('média:', mu, '| DP:', round(sd,1))
print('P(>1050):', round(1-stats.norm.cdf(1050, mu, sd), 4))
```
