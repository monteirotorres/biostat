# Exercícios — Inferência e intervalo de confiança

```python
import numpy as np
import pandas as pd
from scipy import stats
```

## Exercício 3 — IC 95%

```python
a = pd.Series([5,3,4,2,3,4,2,3,4,5])
print('média:', a.mean(), '| s:', round(a.std(),3))
print('IC95%:', np.round(stats.t.interval(0.95, len(a)-1, a.mean(), stats.sem(a)), 2))
```

## Exercício 4 — Moeda viciada (binomial, 20 lançamentos)

```python
p_caudas = stats.binom.cdf(5,20,0.5) + (1-stats.binom.cdf(14,20,0.5))
print('P(<=5 ou >=15):', round(p_caudas, 4), '→ < 5% logo rejeita honesta')
```
