## Poder por simulação (t não pareado)

```python
def poder(n, delta, sigma=1, reps=2000, seed=0):
    rng = np.random.default_rng(seed); r = 0
    for _ in range(reps):
        a = rng.normal(0, sigma, n); b = rng.normal(delta, sigma, n)
        if stats.ttest_ind(a, b).pvalue < 0.05: r += 1
    return r/reps
print('n=17, delta=1:', poder(17, 1))
```
