# Exercícios — Distribuição normal

```python
import numpy as np
import pandas as pd
from scipy import stats
```

## Exercício 1 — Regra 68-95-99,7

```python
for k in [1,2,3]:
    print(f'±{k}σ:', round(stats.norm.cdf(k)-stats.norm.cdf(-k), 4))
```

## Exercício 3 — Elevador (soma de pesos)

```python
for n in [7,6]:
    mu, sd = 70*n, 10*np.sqrt(n)
    print(f'{n} passageiros: P(soma>500) =', round(1-stats.norm.cdf(500, mu, sd), 4))
```

## Exercício 4 — Pressão arterial N(100,10)

```python
print('% entre 80 e 120:', round(stats.norm.cdf(120,100,10)-stats.norm.cdf(80,100,10), 4))
print('faixa 95%:', np.round(stats.norm.interval(0.95,100,10), 1))
print('1% acima de:', round(stats.norm.ppf(0.99,100,10), 1))
```

## Exercício 6 — Garantia
