## Exercício 2 — Salários

```python
sal = pd.Series([2]*1000 + [3]*3000 + [4]*6000)
print('média:', sal.mean(), '| mediana:', sal.median())
```

## Exercício 3 — Faixas de renda (mediana acumulada)

```python
faixas = pd.DataFrame({'faixa':['<1','1-1.5','1.5-3','3-5','5-10','10+'],
                       'pct':[24.1,20.6,28.5,11.7,8.5,6.6]})
faixas['acum'] = faixas['pct'].cumsum()
faixas
```

## Exercício 4 — Tempo de sono de mamíferos

```python
sono = pd.Series([19.9,18.1,18.0,18.0,17.0,16.0,15.8,15.8,14.9,14.6,14.5,14.4,
                  14.3,14.0,13.5,13.1,12.6,12.1,12.1,12.1,11.8,11.4,10.8,10.8,
                  10.6,10.4,10.3,10.3,10.1,9.9,9.7,9.4,8.0,7.8,7.0,6.2,5.5,5.3,
                  3.9,3.9,3.8,3.1,2.9,2.0,1.9])
print('n:', sono.size)
print('média:', round(sono.mean(),2), '| mediana:', sono.median())
print('Q1:', sono.quantile(.25), '| Q3:', sono.quantile(.75),
      '| IQR:', round(stats.iqr(sono),2))
```
