# Exercícios — Precisão e acurácia

```python
import numpy as np
import pandas as pd
from scipy import stats
```

## Exercício 1 — Pipeta

```python
p200 = pd.Series([198.75, 198.83, 198.78, 198.87])
p1000 = pd.Series([996.41, 996.34, 996.48, 996.44])
print('exatidão 200 (viés):', round(p200.mean()-200, 3))
print('exatidão 1000 (viés):', round(p1000.mean()-1000, 3))
print('precisão 200 (DP):', round(p200.std(), 4))
print('precisão 1000 (DP):', round(p1000.std(), 4))
print('CV 200 (%):', round(stats.variation(p200)*100, 4))
print('CV 1000 (%):', round(stats.variation(p1000)*100, 4))
```

## Exercício 5 — Béqueres
