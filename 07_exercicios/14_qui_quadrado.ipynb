# Exercícios — Qui-quadrado

```python
import numpy as np
import pandas as pd
from scipy import stats
```

## Exercício 1 — Dado honesto

```python
obs = [8,11,7,12,15,7]
print('60 lançamentos:', stats.chisquare(obs, [10]*6))
print('600 lançamentos:', stats.chisquare([o*10 for o in obs], [100]*6))
```

## Exercício 2 — Mendel 9:3:3:1

```python
obs = [315,108,101,32]; total = sum(obs)
esp = [total*p for p in [9/16,3/16,3/16,1/16]]
print('Mendel:', stats.chisquare(obs, esp))
```
