# Exercícios — Distribuição binomial

```python
import numpy as np
import pandas as pd
from scipy import stats
```

## Exercício 1 — Falsos positivos acumulados

```python
for k in [3,10,50]:
    print(f'{k} testes: P(>=1 FP) =', round(1-0.95**k, 4))
```

## Exercício 2 — Pôquer (>=3 vitórias em 10, p=1/4)

```python
print(round(1 - stats.binom.cdf(2, 10, 0.25), 4))
```

## Exercício 3 — Mudas

```python
print('P(nenhuma contaminação):', round(0.94**50, 4))
print('P(<10 contaminadas):', round(stats.binom.cdf(9, 50, 0.06), 4))
```

## Exercício 4 — Prova de múltipla escolha

```python
print('10 questões:', round(1-stats.binom.cdf(4,10,0.2), 4))
print('20 questões:', round(1-stats.binom.cdf(9,20,0.25), 4))
```

## Exercício 5 — Controle de qualidade
