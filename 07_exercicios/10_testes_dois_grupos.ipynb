# Exercícios — Testes de dois grupos

```python
import numpy as np
import pandas as pd
from scipy import stats
```

## Exercício 2 — Besouros (Mann-Whitney)

```python
a1 = pd.Series([8,12,15,21,25,44,44,60])
a2 = pd.Series([2,4,5,9,12,17,19])
stats.mannwhitneyu(a1, a2, alternative='two-sided')
```

## Exercício 3 — Halotano x Morfina (t a partir de resumos)

```python
res = stats.ttest_ind_from_stats(66.9,12.2,61, 73.2,14.4,61)
res
```

## Exercício 5 — t não pareado

```python
a1 = pd.Series(range(0,10), dtype=float)
a2 = pd.Series(range(3,13), dtype=float)
stats.ttest_ind(a1, a2)
```
