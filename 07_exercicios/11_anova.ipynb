# Exercícios — ANOVA

```python
import numpy as np
import pandas as pd
from scipy import stats
```

## Exercício 1 — Drogas A, B e placebo

```python
P = [5.4,4.0,7.0,5.8,3.5,7.6,5.5]
A = [6.0,4.8,6.9,6.4,5.5,9.0,6.8]
B = [5.1,3.9,6.5,5.6,3.9,7.0,5.4]
stats.f_oneway(P, A, B)
```

## Exercício 2 — Concurso (5 avaliadores × 6 candidatos)

```python
notas = pd.DataFrame({
    'c1': [9.43,9.20,8.95,8.90,9.10],
    'c2': [9.00,8.88,9.13,8.75,9.15],
    'c3': [8.63,8.65,8.70,8.38,8.95],
    'c4': [8.35,8.18,8.60,7.85,8.35],
    'c5': [7.80,8.33,8.18,7.75,7.68],
    'c6': [7.95,7.55,8.03,8.08,8.08],
})
print('médias dos candidatos:')
print(notas.mean().round(3))
stats.f_oneway(*[notas[c] for c in notas])
```
