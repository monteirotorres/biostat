Teorema de Bayes: P(D|+) = sens·prev / [sens·prev + (1−esp)·(1−prev)]

## Exercício 1 — Rastreio (prev 0,01%)

```python
prev, sens, esp = 0.0001, 0.99, 0.95
print('P(doente|+) =', round(sens*prev/(sens*prev+(1-esp)*(1-prev)), 5))
```

## Exercício 2 — Previsão do tempo

```python
prev, sens, esp = 0.1, 0.8, 0.9
print('P(chuva|previsão+) =', round(sens*prev/(sens*prev+(1-esp)*(1-prev)), 4))
```

## Exercício 3 — AVC e pressão alta

```python
prev, sens, esp = 0.10, 0.40, 0.80
print('P(AVC|PA alta) =', round(sens*prev/(sens*prev+(1-esp)*(1-prev)), 4))
```
