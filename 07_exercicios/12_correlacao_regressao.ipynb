# Exercícios — Correlação e regressão

```python
import numpy as np
import pandas as pd
from scipy import stats
```

## Exercício 1 — Peso x altura

```python
altura = pd.Series([152,157,163,168,173,178,183,188,193])
peso = pd.Series([38,43,64,70,54,79,66,89,68])
r, p = stats.pearsonr(altura, peso)
reg = stats.linregress(altura, peso)
print('r =', round(r,4), '| R² =', round(r**2,4))
print(f'reta: y = {reg.slope:.2f}x + {reg.intercept:.1f}')
```
