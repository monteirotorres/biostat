# Testes não paramétricos

Quando os dados não são normais ou são ordinais.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## Mann–Whitney U (substitui o t independente)

```python
df = sns.load_dataset('penguins').dropna()
a = df[df['species']=='Adelie']['body_mass_g']
g = df[df['species']=='Gentoo']['body_mass_g']
stats.mannwhitneyu(a, g)
```

## Wilcoxon (substitui o t pareado)

```python
antes  = pd.Series([142, 138, 145, 150, 139, 144, 141, 147])
depois = pd.Series([135, 130, 140, 142, 132, 138, 134, 139])
stats.wilcoxon(antes, depois)
```

## Kruskal–Wallis (substitui a ANOVA)

```python
grupos = [g['body_mass_g'].values for _, g in df.groupby('species')]
stats.kruskal(*grupos)
```
