# Teste t pareado

Cada indivíduo é medido **duas vezes** (antes e depois, por exemplo).

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
antes  = pd.Series([142, 138, 145, 150, 139, 144, 141, 147, 152, 137])
depois = pd.Series([135, 130, 140, 142, 132, 138, 134, 139, 144, 130])
stats.ttest_rel(antes, depois)
```

```python
df = pd.DataFrame({'antes': antes, 'depois': depois}).melt(var_name='momento', value_name='PA')
sns.boxplot(data=df, x='momento', y='PA');
```
