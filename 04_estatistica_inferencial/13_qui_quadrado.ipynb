# Teste do qui-quadrado ($\chi^2$)

Testa associação entre **variáveis categóricas**.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
df = sns.load_dataset('titanic')
tabela = pd.crosstab(df['sex'], df['survived'])
tabela
```

```python
chi2, p, dof, esperado = stats.chi2_contingency(tabela)
print(f'chi² = {chi2:.2f}')
print(f'p-valor = {p}')
print(f'graus de liberdade = {dof}')
```

## Visualizando

```python
sns.heatmap(tabela, annot=True, fmt='d', cmap='Blues');
```
