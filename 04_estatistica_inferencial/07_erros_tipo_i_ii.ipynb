# Erros tipo I e tipo II

- **Tipo I** ($\alpha$): rejeitar H₀ quando ela era verdadeira (falso positivo).
- **Tipo II** ($\beta$): não rejeitar H₀ quando ela era falsa (falso negativo).

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## Simulando o erro tipo I

Quando H₀ é verdadeira, esperamos rejeitar H₀ em ~5% dos testes (α=0.05).

```python
import numpy as np
rng = np.random.default_rng(0)
rejeicoes = 0
for _ in range(1000):
    a = rng.normal(100, 15, 30)
    b = rng.normal(100, 15, 30)  # mesma média!
    if stats.ttest_ind(a, b).pvalue < 0.05:
        rejeicoes += 1
rejeicoes / 1000
```

## Poder estatístico (1 − β)

```python
from scipy.stats import norm
n, mu0, mu1, sigma, alpha = 30, 100, 105, 15, 0.05
se = sigma / n**0.5
z_crit = norm.ppf(1 - alpha/2)
poder = 1 - norm.cdf(mu0 + z_crit*se, mu1, se) + norm.cdf(mu0 - z_crit*se, mu1, se)
poder
```
