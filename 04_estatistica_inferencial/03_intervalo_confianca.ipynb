# Intervalos de confiança

Um intervalo de 95% significa que, se repetíssemos o experimento muitas vezes, 95% dos intervalos calculados conteriam o parâmetro verdadeiro.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## Construindo IC para a média

```python
df = sns.load_dataset('penguins').dropna()
amostra = df['body_mass_g'].sample(50, random_state=0)
ic = stats.t.interval(0.95, df=len(amostra)-1, loc=amostra.mean(), scale=stats.sem(amostra))
ic
```

## Visualizando 50 ICs

```python
import numpy as np
verdadeiro = df['body_mass_g'].mean()
fig, ax = plt.subplots(figsize=(6, 8))
for i in range(50):
    s = df['body_mass_g'].sample(30, random_state=i)
    lo, hi = stats.t.interval(0.95, df=len(s)-1, loc=s.mean(), scale=stats.sem(s))
    cor = 'tab:blue' if lo <= verdadeiro <= hi else 'tab:red'
    ax.plot([lo, hi], [i, i], color=cor)
ax.axvline(verdadeiro, color='black', linestyle='--');
```

Os intervalos em vermelho **não** contêm o valor verdadeiro — esperamos que ~5% dos 95%-ICs falhem.
