# Estimação

Usar dados da amostra para estimar parâmetros da população. Pode ser **pontual** (um número) ou **intervalar** (um intervalo).

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
df = sns.load_dataset('penguins').dropna()
x = df['body_mass_g']
```

## Estimativa pontual

```python
x.mean()
```

## Estimativa intervalar (intervalo de confiança 95%)

```python
stats.t.interval(0.95, df=len(x)-1, loc=x.mean(), scale=stats.sem(x))
```
