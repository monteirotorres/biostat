# Teste z para uma amostra

Quando conhecemos o desvio padrão da população.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
import numpy as np
# Glicemia média populacional conhecida: μ=90 mg/dL, σ=12
amostra = pd.Series(stats.norm.rvs(loc=95, scale=12, size=40, random_state=0))
mu_0 = 90
sigma = 12
z = (amostra.mean() - mu_0) / (sigma / np.sqrt(len(amostra)))
p = 2 * (1 - stats.norm.cdf(abs(z)))
z, p
```
