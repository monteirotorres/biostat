# Nível de significância (α)

É o limiar que escolhemos para considerar um resultado 'estatisticamente significativo'.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
alpha = 0.05
p = 0.03
p < alpha  # significativo?
```

## Distribuição da diferença sob H0

```python
import numpy as np
x = np.linspace(-4, 4, 500)
y = stats.norm.pdf(x)
plt.plot(x, y)
z_crit = stats.norm.ppf(1 - 0.025)
plt.axvline( z_crit, color='red')
plt.axvline(-z_crit, color='red')
plt.fill_between(x, y, where=(x > z_crit) | (x < -z_crit), color='red', alpha=0.3);
```
