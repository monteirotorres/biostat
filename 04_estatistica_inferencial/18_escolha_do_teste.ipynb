# Como escolher o teste

Um guia prático: cada situação e a função correspondente em `scipy.stats`.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## Comparar grupos

```python
import numpy as np
rng = np.random.default_rng(0)
a = rng.normal(10, 2, 20); b = rng.normal(11, 2, 20); c = rng.normal(12, 2, 20)
# 1 amostra vs referência
print('t 1 amostra:', stats.ttest_1samp(a, 10))
# 2 grupos independentes (Welch)
print('t Welch:', stats.ttest_ind(a, b, equal_var=False))
# 2 grupos pareados
print('t pareado:', stats.ttest_rel(a, b))
# 3+ grupos
print('ANOVA:', stats.f_oneway(a, b, c))
```

## Versões não-paramétricas

```python
print('Wilcoxon sinal:', stats.wilcoxon(a - 10))
print('Mann-Whitney:', stats.mannwhitneyu(a, b))
print('Wilcoxon pareado:', stats.wilcoxon(a, b))
print('Kruskal-Wallis:', stats.kruskal(a, b, c))
```

## Associação entre quantitativas

```python
x = rng.normal(0, 1, 40); y = 0.6*x + rng.normal(0, 1, 40)
print('Pearson:', stats.pearsonr(x, y))
print('Spearman:', stats.spearmanr(x, y))
print('regressão:', stats.linregress(x, y))
```

## Variáveis categóricas

```python
tab = np.array([[30, 10], [12, 28]])
print('qui-quadrado:', stats.chi2_contingency(tab)[:2])
print('Fisher:', stats.fisher_exact(tab))
```

## Pós-testes da ANOVA

Para Tukey HSD use `scipy.stats.tukey_hsd`; para Dunnett, `scipy.stats.dunnett`.

```python
print('Tukey HSD:')
print(stats.tukey_hsd(a, b, c))
print('Dunnett (b, c vs controle a):')
print(stats.dunnett(b, c, control=a))
```
