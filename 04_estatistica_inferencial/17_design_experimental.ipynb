# Design experimental e potência

Simulando o poder estatístico de um teste t.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## Poder por simulação

Fração de experimentos em que rejeitamos H₀ quando há efeito real.

```python
def poder_simulado(n, delta, sigma=1, alpha=0.05, reps=2000, seed=0):
    rng = np.random.default_rng(seed)
    rejeicoes = 0
    for _ in range(reps):
        a = rng.normal(0, sigma, n)
        b = rng.normal(delta, sigma, n)
        if stats.ttest_ind(a, b).pvalue < alpha:
            rejeicoes += 1
    return rejeicoes / reps

poder_simulado(n=17, delta=1)
```

## Curva de poder vs. tamanho da amostra

```python
ns = [5, 10, 17, 25, 40, 60, 100]
poderes = [poder_simulado(n, delta=1) for n in ns]
plt.plot(ns, poderes, 'o-')
plt.axhline(0.8, color='red', linestyle='--', label='80%')
plt.xlabel('n por grupo'); plt.ylabel('poder'); plt.legend();
```

## Efeito do tamanho do efeito (δ)

```python
deltas = [0.2, 0.5, 0.8, 1.0, 1.5, 2.0]
poderes = [poder_simulado(n=20, delta=d) for d in deltas]
plt.plot(deltas, poderes, 'o-')
plt.axhline(0.8, color='red', linestyle='--')
plt.xlabel('tamanho do efeito (δ/σ)'); plt.ylabel('poder');
```
