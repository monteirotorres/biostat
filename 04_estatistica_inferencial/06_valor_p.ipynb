# Valor-p

Probabilidade de obter dados **tão ou mais extremos** que os observados, supondo que H₀ seja verdadeira.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## Exemplo: moeda joga 100 vezes, dá 60 caras

```python
# Teste bilateral
stats.binomtest(60, n=100, p=0.5)
```

Como o p-valor é maior que 0,05, **não rejeitamos** H₀ no nível de 5%.

## Teste t para uma amostra

```python
import numpy as np
rng = np.random.default_rng(0)
amostra = pd.Series(rng.normal(102, 15, size=30))
stats.ttest_1samp(amostra, popmean=100)
```
