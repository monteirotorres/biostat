# Teste t para uma amostra

Compara a média de uma amostra com um valor de referência. Usa-se quando $\sigma$ é desconhecido.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
import numpy as np
amostra = pd.Series(stats.norm.rvs(loc=102, scale=14, size=25, random_state=0))
stats.ttest_1samp(amostra, popmean=100)
```

## E se o p-valor for unilateral?

```python
stats.ttest_1samp(amostra, popmean=100, alternative='greater')
```
