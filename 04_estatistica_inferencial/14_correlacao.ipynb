# Correlação

Mede a força e direção da associação linear entre **duas variáveis numéricas**.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
df = sns.load_dataset('penguins').dropna()
stats.pearsonr(df['flipper_length_mm'], df['body_mass_g'])
```

```python
sns.scatterplot(data=df, x='flipper_length_mm', y='body_mass_g', hue='species');
```

## Matriz de correlação

```python
df.select_dtypes('number').corr()
```

```python
sns.heatmap(df.select_dtypes('number').corr(), annot=True, cmap='coolwarm', center=0);
```

## Spearman — correlação não paramétrica

```python
stats.spearmanr(df['flipper_length_mm'], df['body_mass_g'])
```
