# ANOVA — análise de variância

Compara as médias de **três ou mais grupos**.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
df = sns.load_dataset('penguins').dropna()
grupos = [g['body_mass_g'].values for _, g in df.groupby('species')]
stats.f_oneway(*grupos)
```

```python
sns.boxplot(data=df, x='species', y='body_mass_g');
```

Se p < 0.05, **alguma** das médias é diferente. Para saber qual, fazemos comparações pareadas.
