# Hipótese nula (H₀) e alternativa (H₁)

- **H₀**: 'não há diferença' / 'nada está acontecendo'
- **H₁**: 'há diferença' / 'o efeito existe'

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## Exemplo: novo remédio reduz pressão arterial?

- H₀: a pressão média **não** muda com o remédio.
- H₁: a pressão média muda com o remédio.

```python
antes  = pd.Series([142, 138, 145, 150, 139, 144, 141, 147])
depois = pd.Series([135, 130, 140, 142, 132, 138, 134, 139])
stats.ttest_rel(antes, depois)
```

`pvalue` muito pequeno → temos evidência para **rejeitar H₀** em favor de H₁.
