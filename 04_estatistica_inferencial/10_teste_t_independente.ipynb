# Teste t para duas amostras independentes

Compara as médias de **dois grupos diferentes**.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
df = sns.load_dataset('penguins').dropna()
adelie  = df[df['species']=='Adelie']['body_mass_g']
gentoo  = df[df['species']=='Gentoo']['body_mass_g']
stats.ttest_ind(adelie, gentoo, equal_var=False)  # Welch
```

```python
sns.boxplot(data=df[df['species'].isin(['Adelie','Gentoo'])],
            x='species', y='body_mass_g');
```
