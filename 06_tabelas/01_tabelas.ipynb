# Tabelas estatísticas com scipy

Como obter qualquer valor crítico ou área sem consultar tabelas impressas.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## Normal padrão (tabela Z)

```python
print('P(Z < 1.96) =', stats.norm.cdf(1.96))
print('z para 95% bilateral =', stats.norm.ppf(0.975))
```

## Valores críticos t

```python
for df in [5, 10, 30, 100]:
    print(f'gl={df}: t* 95% = {stats.t.ppf(0.975, df):.3f}')
```

## Valores críticos qui-quadrado

```python
for df in [1, 5, 10]:
    print(f'gl={df}: chi2* (5%) = {stats.chi2.ppf(0.95, df):.3f}')
```

## Valores críticos F (ANOVA)

```python
print('F* (5%, 2, 27) =', stats.f.ppf(0.95, 2, 27))
```

## Construindo uma tabela Z em DataFrame

```python
z = np.round(np.arange(0, 3.5, 0.1), 1)
cols = np.round(np.arange(0, 0.1, 0.01), 2)
tab = pd.DataFrame({c: stats.norm.cdf(z + c) for c in cols}, index=z).round(4)
tab.head(10)
```
