# Construindo a distribuição qui-quadrado

χ² com k graus de liberdade = soma de k normais padrão ao quadrado.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
rng = np.random.default_rng(0)
k = 4
amostras = (rng.normal(size=(20000, k))**2).sum(axis=1)
sns.histplot(amostras, stat='density', bins=60)
x = np.linspace(0, 25, 300)
plt.plot(x, stats.chi2.pdf(x, df=k), 'r', label=f'chi2 teórica (df={k})')
plt.legend();
```

O histograma das somas de quadrados coincide com `stats.chi2`. Média ≈ k, variância ≈ 2k.

```python
print('média simulada:', amostras.mean(), ' teórica:', k)
print('variância simulada:', amostras.var(), ' teórica:', 2*k)
```

## Efeito dos graus de liberdade

```python
x = np.linspace(0, 30, 300)
for k in [1, 3, 6, 10]:
    plt.plot(x, stats.chi2.pdf(x, df=k), label=f'df={k}')
plt.legend(); plt.title('Distribuição qui-quadrado');
```
