# De onde vêm t, F e χ²

Vamos ver, por simulação, que essas distribuições nascem da amostragem de variáveis normais.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

## A média amostral é normal (TLC)

```python
medias = [stats.norm.rvs(size=30, random_state=i).mean() for i in range(2000)]
sns.histplot(medias, stat='density', bins=40)
x = np.linspace(-1, 1, 200)
plt.plot(x, stats.norm.pdf(x, 0, 1/np.sqrt(30)), 'r')
plt.title('Distribuição da média amostral');
```
