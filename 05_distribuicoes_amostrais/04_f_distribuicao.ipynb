# Construindo a distribuição F

F = razão entre duas variâncias (duas qui-quadrado, cada uma sobre seus gl).

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
rng = np.random.default_rng(0)
d1, d2 = 5, 10
fs = []
for _ in range(20000):
    a = rng.normal(size=d1+1)
    b = rng.normal(size=d2+1)
    fs.append(a.var(ddof=1) / b.var(ddof=1))
fs = np.array(fs)
```

```python
sns.histplot(fs, stat='density', bins=80)
x = np.linspace(0, 6, 300)
plt.plot(x, stats.f.pdf(x, d1, d2), 'r', label=f'F({d1},{d2})')
plt.xlim(0, 6); plt.legend();
```

## Relação t² = F(1, k)

```python
k = 10
t = stats.t.rvs(df=k, size=20000, random_state=0)
print('quantil 95% de t²:', np.quantile(t**2, 0.95))
print('quantil 95% de F(1,k):', stats.f.ppf(0.95, 1, k))
```
