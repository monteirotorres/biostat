# Construindo a distribuição t de Student

t surge quando padronizamos a média usando o desvio padrão **estimado** (s), não o populacional (σ).

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats

sns.set_theme(style="whitegrid")
```

```python
rng = np.random.default_rng(0)
n = 5
ts = []
for _ in range(20000):
    x = rng.normal(0, 1, n)
    ts.append(x.mean() / (x.std(ddof=1) / np.sqrt(n)))
ts = np.array(ts)
```

```python
sns.histplot(ts, stat='density', bins=120)
x = np.linspace(-6, 6, 400)
plt.plot(x, stats.t.pdf(x, df=n-1), 'r', label=f't (df={n-1})')
plt.plot(x, stats.norm.pdf(x), 'g--', label='normal')
plt.xlim(-6, 6); plt.legend();
```

Repare nas **caudas mais pesadas** que a normal (verde). Com n grande, t → normal.

```python
x = np.linspace(-4, 4, 400)
for df in [1, 3, 10, 30]:
    plt.plot(x, stats.t.pdf(x, df=df), label=f'df={df}')
plt.plot(x, stats.norm.pdf(x), 'k--', label='normal')
plt.legend();
```
